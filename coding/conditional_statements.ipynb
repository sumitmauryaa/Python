```python
# Q1. Accept two number and print the greatest between them 

first_number = int(input("Enter the first number :"))
second_number = int(input("Enter the second number :"))
if first_number > second_number:
    print("First number is the greatest number :", first_number)
elif first_number == second_number:
    print("Both numbers are equal !")
else:
    print("Second number is the greatest number :", second_number)
```

```python
# Q2. Accept the gender from the user as char and print the respective greeting message
#     Ex - Good Morning Sir (on the basis of gender)


gender = input("Enter Your gender 'M' for male OR 'F' for female :")

if gender == 'M' or gender == 'm':
    print("Good Morning Sir !")
elif gender == 'F' or gender == 'f':
    print("Good Morning Mam !")
else: 
    print("You are entered wrong character")
```

```python
# Q3. Accept an integer and check whether it is an even number or odd.

num = int(input("Enter the number :"))
if num %2 == 0:
    print("This  is Even Number.")
else:
    print("This is Odd number !!")
```

```python
# Q4. Accept name and age from the user. Check if the user is a valid voter or not.
#     Ex- “hello shery you are a valid voter”

name = input("Enter the Name :")
age = int(input("Enter you current Age :"))
if age >=18:
    print(f"hello {name} you are a valid voter")
else:
    print(f"Hello {name}, You are not eligible to vote !!")
```

```python
# Q5. Accept a year and check if it a leap year or not (google to find out what is a leap year)

Year = int(input("Enter the year to check Leap year or not :  "))
if (Year % 4 == 0 and Year % 100 != 0) or (Year % 400 == 0):
    print("This  is a leap year ")
else:
    print("This is not a Leap year")
```

```python
#     Q6.     @ take the input of temperature in celsius
#             @ Below 0°C → "Freezing Cold 
#             @ 0°C to 10°C → "Very Cold 
#             @ 10°C to 20°C → "Cold 
#             @ 20°C to 30°C → "Pleasant 
#             @ 30°C to 40°C → "Hot 
#             @ Above 40°C → "Very Hot                        

celsius = int(input("Enter the celsius :"))
if celsius <=0:
    print("Freezing Point ")
elif celsius >0 and celsius<=10:
    print("Very Cold")
elif celsius >10 and celsius<=20:
    print("Cold")
elif celsius >20 and celsius<=30:
    print("Pleasant")
elif celsius >30 and celsius<=40:
    print("Hot")
elif celsius >40 :
    print("Very Hot")
else:
    print("Wrong Input")
```
