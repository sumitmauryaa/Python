```python
# 1. Accept an integer and Print hello world n times 

n = int(input("Enter the number to print n times :"))
for i in range(1,n):
    print("Hello World",i)
```

```python
# 2. Print natural number up to n 
n = int(input("Enter the number to print natural number upto your number :"))
for i in range (1,n+1):
    print(i)
```

```python
# 3. Reverse for loop. Print n to 1 

n = int(input("Enter the number :"))
for i in range(n,0,-1):
    print(i, end=" ")
```

```python
# 4. Take a number as input and print its table

n = int(input("Enter the number : "))
for i in range(1,11):
    print(f"{n} X {i} = ",n*i)
```

```python
# 5. Sum up to n terms 

n = int(input("Enter the number :"))
sum = 0
for i in range(0 , n+1):
    sum = sum + i
print(sum)
```

```python
# 6. Factorial of a number 
n = int(input("Enter the number :"))
fact = 1
for i in range(1,n+1):
    fact = fact * i
print(fact)
```

```python
# 7. Print the sum of all even & odd numbers in a range separately 

n = int(input("Enter the number :"))
even = 0
odd = 0
for i in range(1, n+1):
    if i % 2 == 0 :
        even = even + i
    elif i %2 != 0:
        odd = odd + i
print("Sum of Even Numbers :",even)
print("Sum of Odd Number : ",odd)
```

```python
# 8. Print all the factors of a number 

n = int(input("Enter the number :"))
for i in range(1,n+1):
    if n % i ==0:
        print(i)
```
